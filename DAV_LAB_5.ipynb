```python
import pandas as pd
import sqlite3
from sqlalchemy import create_engine, text
```

```python
data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Kimi"],
    "Department": ["CSE", "CSE","IT", "CSE", "IT"],
    "Marks": [89,99,75,56,78],
    "City": ["Nagpur","Balaghat","Pune", "Mumbai","Bhopal"]
}

df = pd.DataFrame(data)
df
```

```python
df.to_excel("student.xlsx", index=False)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.dtypes
```

Part B : Sqlite database using splite3

```python
conn = sqlite3.connect("student.db")
```

```python
df.to_sql(
    "students",
    conn,
    if_exists="replace",
    index=False
)
```

```python
query = """
SELECT * FROM students;
"""
```

```python
result = pd.read_sql_query(query, conn)
result
```

```python
query = """
SELECT Student_ID, Name, Marks
FROM students
WHERE Marks> 80;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT AVG(Marks) AS Average_Marks
FROM students;
"""

pd.read_sql_query(query, conn)
```

```python

query = """
SELECT Department, AVG(Marks) AS Average_Marks
FROM students
GROUP BY Department;
"""

pd.read_sql_query(query, conn)
```

Part C : SQLite using SqlAlchemy

```python
engine = create_engine("sqlite:///student_sqlaichemy.db")
```

```python
df.to_sql(
    "students",
    engine,
    if_exists="replace",
    index=False
)
```

```python
query = """
SELECT * FROM students;
"""

result = pd.read_sql_query(query, engine)
result
```

```python
with engine.connect() as connection:
    result = connection.execute(text("SELECT * FROM students WHERE Marks >80"))
    for row in result:
        print(row)
```
