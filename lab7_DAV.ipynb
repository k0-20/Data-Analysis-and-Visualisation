```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Lab7_Dataset - Sheet1.csv")
```

```python
#Experiment 1: Basic Line Plot

plt.figure(figsize=(10, 5))
plt.plot(df["Month"],
         df["SALES"],
         marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
#Experiment 2: Customized Line Plot

plt.figure(figsize=(12, 6))
plt.plot(    df["Month"],
         df["SALES"],
             marker="o",
             linewidth=2,
             label="Sales")
plt.title(    "Monthly Sales Trend",
          fontsize=16)
plt.xlabel("Month")
plt.ylabel("SALES")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
#Experiment 3: Multiple Trend Lines

plt.figure(figsize=(12, 6))
plt.plot(    df["Month"],
         df["SALES"],
             marker="o",
             label="SALES")
plt.plot(    df["Month"],
         df["Profit"],
             marker="s",
             label="Profit")
plt.title("Monthly Sales and Profit")
plt.xlabel("Month")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
#Experiment 4: Seaborn Line Plot
plt.figure(figsize=(12, 6))
sns.lineplot(    data=df,
             x="Month",
                 y="SALES",
                 marker="o")
plt.title("Monthly Sales Using Seaborn")
plt.xlabel("Month")
plt.ylabel("SALES")
plt.show()

```

```python
#Experiment 5: Annotating Important Observations

max_sales = df["SALES"].max()
max_month = df.loc[    df["SALES"].idxmax(),
                   "Month"]
print(max_month, max_sales)

plt.figure(figsize=(12, 6))
plt.plot(    df["Month"],
         df["SALES"],
             marker="o")
plt.annotate(    f"Highest: {max_sales}",
             xy=(max_month, max_sales),
                 xytext=(-60, -40),
                 textcoords="offset points",
                 arrowprops=dict(arrowstyle="->"))
plt.title("Sales Trend with Maximum Sales Highlighted")
plt.xlabel("Month")
plt.ylabel("SALES")
plt.grid(True)
plt.show()
```

```python
#Experiment 6: Scatter Plot

plt.figure(figsize=(8, 6))
plt.scatter(    df["Advertising"],
            df["SALES"])
plt.title("Advertising vs Sales")
plt.xlabel("Advertising")
plt.ylabel("SALES")
plt.grid(True)
plt.show()

```

```python
#Experiment 7: Scatter Plot Using Seaborn
plt.figure(figsize=(8, 6))
sns.scatterplot(    data=df,
                x="Advertising",
                    y="SALES")
plt.title("Advertising vs Sales")
plt.show()
```

```python
#Experiment 8: Scatter Plot with Color
#Use customer count to provide additional visual information.
plt.figure(figsize=(9, 6))
sns.scatterplot(    data=df,
                x="Advertising",
                    y="SALES",
                    hue="Customers",
                    palette="viridis")
plt.title("Advertising vs Sales by Customer Count")
plt.show()




```

```python
#Experiment 9: Scatter Plot with Variable Size
#Use customer count to control point size.
plt.figure(figsize=(9, 6))
sns.scatterplot(    data=df,
                x="Advertising",
                    y="SALES",
                    hue="Customers",
                    size="Customers",
                    sizes=(50, 400),
                    palette="viridis")
plt.title("Advertising vs Sales")
plt.xlabel("Advertising")
plt.ylabel("SALES")
plt.show()
```

```python
#Experiment 10: Scatter Plot with Trend Line
#Use Seaborn's regression functionality.
plt.figure(figsize=(9, 6))
sns.regplot(    data=df,
            x="Advertising",
                y="SALES")
plt.title("Advertising vs Sales with Trend Line")
plt.xlabel("Advertising")
plt.ylabel("SALES")
plt.show()


```

```python
#Experiment 11: Regression Plot with Confidence Interval
plt.figure(figsize=(9, 6))
sns.regplot(    data=df,
            x="Advertising",
                y="SALES",
                ci=95)
plt.title("Advertising vs Sales – Regression Analysis")
plt.show()

```

```python
#Experiment 12: Calculate Correlation
correlation = df["Advertising"].corr(df["SALES"])
print("Correlation:", correlation)

correlation_matrix = df.corr(numeric_only=True)
print(correlation_matrix)

plt.figure(figsize=(8, 6))
sns.heatmap(    correlation_matrix,
            annot=True,
                cmap="coolwarm",
                fmt=".2f")
plt.title("Correlation Matrix")
plt.show()


```

```python
#Experiment 13: Identifying an Important Observation
#Find the month with the highest number of customers.
max_customers = df["Customers"].max()
row = df.loc[    df["Customers"].idxmax()]
print(row)

plt.figure(figsize=(9, 6))
sns.scatterplot(    data=df,
                x="Advertising",
                    y="SALES")
plt.annotate(    f"{row['Month']}",
             xy=(row["Advertising"],
                 row["SALES"]),
                 xytext=(10, 10),
                 textcoords="offset points",
                 arrowprops=dict(arrowstyle="->"))
plt.title("Advertising vs Sales with Annotation")
plt.show()
```
